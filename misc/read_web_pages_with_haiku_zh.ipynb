# 用 Claude 3 Haiku 总结网页内容
在本示例中，我们将学习如何在给定 URL 的情况下抓取网页内容，然后使用 Anthropic 的 Claude API 生成页面内容的摘要。

我们先安装 Anthropic 库。

## 设置
首先，安装必要的库，并用 API 密钥设置我们的 Anthropic 客户端。

```python
# Install the necessary libraries
%pip install anthropic
```

```python
# Import the required libraries
from anthropic import Anthropic

# Set up the Claude API client
client = Anthropic()
MODEL_NAME = "claude-haiku-4-5"
```

## 第 1 步：抓取网页内容
首先，我们需要使用提供的 URL 抓取网页内容。这里我们使用 requests 库。

```python
import requests

url = "https://en.wikipedia.org/wiki/96th_Academy_Awards"
response = requests.get(url, timeout=30)

if response.status_code == 200:
    page_content = response.text
else:
    print(f"Failed to fetch the web page. Status code: {response.status_code}")
    exit(1)
```

## 第 2 步：为 Claude 准备输入
接下来，我们为 Claude API 准备输入。我们创建一条消息，其中包含页面内容，以及一个要求 Claude 总结它的提示词。

```python
prompt = (
    f"<content>{page_content}</content>Please produce a concise summary of the web page content."
)

messages = [{"role": "user", "content": prompt}]
```

## 第 3 步：生成摘要
现在，我们调用 Haiku 生成网页内容的摘要。

```python
response = client.messages.create(model="claude-haiku-4-5", max_tokens=1024, messages=messages)

summary = response.content[0].text
print(summary)
```
